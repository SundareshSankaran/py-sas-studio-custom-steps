# py-sas-studio-custom-steps

### This notebook walks you through the main operations offered by the [py-sas-studio-custom-steps](https://pypi.org/project/py-sas-studio-custom-steps/) package which helps you programmatically create and modify a SAS Studio Custom Step.

### Imports

##### Assuming you have installed py-sas-studio-custom-steps in a virtual environment for testing.

```python
from py_sas_studio_custom_steps import CustomStep
import os
```

### Set path to folder containing example steps and test outputs

##### Most of these are part of this repository and an open-source SAS repository, SAS Studio Custom Steps at https://github.com/sassoftware/sas-studio-custom-steps

```python
test_path = os.path.join(os.getcwd(),"..","tests")
existing_step = os.path.join(test_path,"DuckDB - Extract Parquet Metadata.step")
extracted_sas_file_local_name = os.path.join(test_path,"DuckDB - Extract Parquet Metadata.sas")
extracted_ui_local_name = os.path.join(test_path,"DuckDB - Extract Parquet Metadata components.json")

existing_step_on_gh = "https://github.com/sassoftware/sas-studio-custom-steps/blob/main/DuckDB%20-%20Aggregate%20Parquets/DuckDB%20-%20Aggregate%20Parquets.step"
extracted_sas_file_gh_name = os.path.join(test_path,"DuckDB - Aggregate Parquets.sas")
extracted_ui_gh_name = os.path.join(test_path,"DuckDB - Aggregate Parquets components.json")

example_sas_code = os.path.join(test_path,"example_sas_code.sas")
example_ui_config = os.path.join(test_path,"example_components.json")

example_README_file = os.path.join(test_path,"README.md")
```

### Step 1: Create a Custom Step instance

```python
# A new custom step object.  The __dict__ representation shows its structure. 
cs = CustomStep()
print(type(cs))
print("-----")
print(cs.__dict__)
```

#### Alternatively, create an instance based on an existing custom step

```python
# This custom step exists locally
cs_existing = CustomStep(existing_step)
cs_existing.name
```

```python
# This custom step exists on GitHub and is now contained in the cs_existing custom step object.
cs_existing_gh = CustomStep(existing_step_on_gh)
cs_existing_gh.name
```

### Let's extract the SAS program out of a custom step object

```python
# From our empty custom step
cs.extract_sas_program()
```

```python
# From our custom step based on a step from the GitHub repo
cs_existing_gh.extract_sas_program()
```

```python
# Example of writing the extracted sas program to a new local SAS program file

with open(extracted_sas_file_local_name,"w",encoding="utf-8") as f:
    f.write(cs_existing.extract_sas_program())
```

### Let's now extract the UI (a JSON structure) from the custom step object

```python
# From our empty custom step
cs.extract_ui()
```

```python
# From our object based on a local custom step
cs_existing.extract_ui()
```

```python
# Example of writing the extracted UI to a new local JSON file

with open(extracted_ui_local_name,"w",encoding="utf-8") as f:
    f.write(cs_existing.extract_ui())
    
```

### You can even extract the SAS program and UI from a custom step in a GitHub repository
##### The "empty" cs CustomStep object you created earlier serves the purpose of a via media

```python
# TEst of extract sas program from GitHub repo

with open(extracted_sas_file_gh_name,"w",encoding="utf-8") as f:
    f.write(cs.extract_sas_program(existing_step_on_gh))

# TEst of extract UI from GitHub repo

with open(extracted_ui_gh_name,"w",encoding="utf-8") as f:
    f.write(cs.extract_ui(existing_step_on_gh))
```

### Now, let's add a SAS program

```python
# Test of attach_sas_program
cs = CustomStep()
rc = cs.attach_sas_program(example_sas_code)

print(rc)
print("=============")
print(cs.__dict__)
```

### Let's also attach a UI

```python
# Test of attach_ui

rc = cs.attach_ui(ui_json_file= example_ui_config)

print(rc)
print("=============")
print(cs.__dict__)
```

### Let's start documenting this step.  Add an About Page

```python
rc = cs.add_about_page()

print(rc)
print("=============")
print(cs.__dict__)
```

```python
# GEnerate a README from this custom step
user_prompt = """
Generate a README file with very short description of what this step does.  Affix a version number in semantic versioning (based on step's UI config) and note that 
the author's name is Sundaresh Sankaran (sundaresh.sankaran@sas.com). Use today's date for creation.
"""
cs.generate_readme(user_prompt, example_README_file)
```

# Demo - Save your Custom Step's SAS Program 
## (and your UI configuration too if you prefer)

Refer here for details: [extract_ui and extract_sas_program should save to file](https://github.com/SundareshSankaran/py-sas-studio-custom-steps/issues/24)

### Import the CustomStep class

```python
from py_sas_studio_custom_steps import CustomStep
```

### This custom step was randomly chosen.
### You can choose a custom step either from a URL or on your file system.

```python
cs = CustomStep("https://github.com/sassoftware/sas-studio-custom-steps/blob/main/Git%20-%20List%20Local%20Repo%20Changes/Git%20-%20List%20Local%20Repo%20Changes.step")
```

### Identify a location where you want to save the program / UI configuration to.

```python
import os

program_path=os.path.join(os.getcwd(),f"{cs.name}.sas")
ui_config_path=os.path.join(os.getcwd(),f"{cs.name}.json")
```

### Extract SAS Program

```python
cs.extract_sas_program(program_path=program_path)
```

-[X] Success
-[ ] Failure

### Extract UI Configuration

```python
cs.extract_ui(ui_config_path=ui_config_path)
```

-[X] Success
-[ ] Failure
